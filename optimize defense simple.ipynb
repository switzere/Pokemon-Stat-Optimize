```python
def calc_effectiveness(attacking_type, defending_types):
    effectiveness = 1
    for defending_type in defending_types:
        effectiveness *= typeEffectivenessChart[attacking_type][defending_type]
    return effectiveness
```

```python
#"Unless otherwise specified, all divisions and multiplications past the initial base damage calculation are rounded to the nearest integer if the result is not an integer (rounding down at 0.5)."

def damage(defense_evs, base_defense=100, def_mult=1, atk=150, power=100, level=50, bonus=[1]):
    calculated_defense = math.floor((base_defense + defense_evs + 20) * def_mult)

    raw_damage = math.floor((math.floor( math.floor( (2 * level / 5) + 2) * power * ( (atk) / calculated_defense) ) / 50) + 2 )
    raw_damage_unrounded = (( ( (2 * level / 5) + 2) * power * ( (atk) / calculated_defense) ) / 50) + 2

    for b in bonus:
        raw_damage = math.floor(raw_damage * b)
        raw_damage_unrounded = raw_damage_unrounded * b

    return raw_damage, raw_damage_unrounded

def damage_percent(defense_evs, health_evs, base_health=100, base_defense=100, total_evs=66, def_mult=1, atk=150, power=100, level=50, bonus=[1]):
    if (defense_evs + health_evs) > total_evs:  # Constraint check
        print("EV distribution exceeds total EV limit.")
        return None, None  # Invalid solution
    
    calculated_health = base_health + health_evs + 75
    raw_damage, raw_damage_unrounded = damage(defense_evs, base_defense, def_mult, atk, power, level, bonus)

    return raw_damage / calculated_health * 100, raw_damage
```
